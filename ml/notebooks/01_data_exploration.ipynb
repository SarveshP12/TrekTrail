# 01 - Data Exploration

This notebook explores the raw GPS tracks from the GeoLife dataset to gather insights about speeds, altitudes, and trajectory shapes before feature engineering.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import glob

# GeoLife data directory
data_dir = os.path.abspath(os.path.join('..', 'data', 'raw', 'geolife', 'Data'))

print('Scanning for GeoLife Trajectory data...')

# Load a sample trajectory to visualize
user_dirs = sorted(glob.glob(os.path.join(data_dir, '*')))
if user_dirs:
    first_user_traj_dir = os.path.join(user_dirs[0], 'Trajectory')
    traj_files = sorted(glob.glob(os.path.join(first_user_traj_dir, '*.plt')))
    
    if traj_files:
        file_path = traj_files[0]
        print(f"Loading sample file: {file_path}")
        
        # GeoLife PLT format: skip 6 rows
        # Columns: Lat, Lon, 0, Alt, NumDays, Date, Time
        df = pd.read_csv(file_path, skiprows=6, header=None,
                         names=['lat', 'lon', 'zero', 'alt', 'num_days', 'date', 'time'])
        
        # Create a proper datetime index
        df['datetime'] = pd.to_datetime(df['date'] + ' ' + df['time'])
        df.set_index('datetime', inplace=True)
        
        # 1. Plot the GPS Trajectory Map
        plt.figure(figsize=(8, 6))
        plt.plot(df['lon'], df['lat'], marker='o', markersize=2, linestyle='-', alpha=0.5, color='blue')
        plt.title('Raw GPS Trajectory Exploration (Lat/Lon Map)')
        plt.xlabel('Longitude')
        plt.ylabel('Latitude')
        plt.grid(True)
        plt.show()
        
        # 2. Plot Altitude profile over time
        plt.figure(figsize=(10, 4))
        plt.plot(df.index, df['alt'], color='green')
        plt.title('Altitude Profile over Time')
        plt.xlabel('Time')
        plt.ylabel('Altitude (feet)')
        plt.grid(True)
        plt.show()
    else:
        print("No trajectory files found for the first user. Please check the dataset structure.")
else:
    print("GeoLife Data folder not found or empty. Please ensure 'download_geolife.py' completed successfully.")
```
